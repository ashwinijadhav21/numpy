***ARITHMETIC OPERATOR***

```python
a1 = np.array([12,34,45,67,89])
a2 = np.array([34,45,6,89,31])

print(a1+a2)
print(a1-a2)
print(a1*a2)
print(a1/a2)
print(a1//a2)
```

```python
#Broadcasting with matrix
a=np.arange(1,31).reshape(6,5)
print(a+2)
```

***SHALLOW COPY***

```python
a=np.arange(1,11)
b=a.view()
b[0]=99
print(a)
```

***DEEP COPY***

```python
a=np.arange(1,11)
b=a.copy()
b[0]=99
print(a)
print(b)
```

***MATRIX OPERATION***

```python
#Matrix multiplaction using dot product

a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])

result = np.dot(a,b)
print(result)
```

```python
#Element wise multiplication

a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])

print(a*b)
```

***MATRIX TRANSPOSE***

```python
a=np.array([[1,2],[3,4]])
b=a.T
print(b)
```

***STACKING ARRAYS***

```python
#Vertical stacking

a=np.array([1,2,3])
b=np.array([4,5,6])

result=np.vstack((a,b))
print(result)
```

```python
#Horizontal stacking

a=np.array([1,2,3])
b=np.array([4,5,6])

result = np.hstack((a,b))
print(result)
```

```python
#column stacking

a=np.array([1,2,3])
b=np.array([4,5,6])

result = np.column_stack((a,b))
print(result)
```

***SPLITTING ARRAYS***

```python
a=np.arange(16).reshape(4,4)
a
```

```python
#Horizontal splitting

a=np.arange(16).reshape(4,4)
result = np.hsplit(a,2)
print(result)
```

```python
#Vertical splitting

a=np.arange(16).reshape(4,4)
result = np.vsplit(a,4)
print(result)
```
